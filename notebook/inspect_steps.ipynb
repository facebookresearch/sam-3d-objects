```python
import glob
import os

import matplotlib.pyplot as plt
import torch

# Notebook lives in notebook/, steps dir is one level up
STEPS_DIR = os.path.join("..", "sam3d_objects", "ss_steps")

pt_files = sorted(glob.glob(os.path.join(STEPS_DIR, "step_*.pt")))
print(f"Found {len(pt_files)} step files:")
for f in pt_files:
    print(" ", os.path.basename(f))
```

```python
# Load a single step and print everything in it
data = torch.load(pt_files[0], map_location="cpu", weights_only=False)

for k, v in data.items():
    if hasattr(v, 'shape'):
        print(f"  {k:25s} shape={str(v.shape):25s} dtype={v.dtype}")
    else:
        print(f"  {k:25s} value={v}")
```

```python
fig, axes = plt.subplots(len(pt_files), 3, figsize=(9, 3 * len(pt_files)))

for row, pt_path in enumerate(pt_files):
    data = torch.load(pt_path, map_location="cpu", weights_only=False)
    occ = data["ss_grid"][0, 0]   # (64, 64, 64)
    t   = float(data["t_step"])
    mid = 32

    for col, (slic, title) in enumerate([
        (occ[mid],      f"t={t:.2f}  XY"),
        (occ[:, mid],   "XZ"),
        (occ[:, :, mid],"YZ"),
    ]):
        axes[row, col].imshow(slic.numpy(), cmap="RdBu_r")
        axes[row, col].set_title(title)
        axes[row, col].axis("off")

plt.tight_layout()
plt.savefig("outputs/occ_over_time.png", dpi=100)
plt.show()
```
